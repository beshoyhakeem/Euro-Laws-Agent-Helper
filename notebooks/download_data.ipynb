````python
 # for progress bar (install with: pip install tqdm)

# Assuming your dataframe is called 'law_df'
# First, create a dictionary mapping CELEX to act_raw_text for faster lookup
celex_to_text = dict(zip(law['CELEX'], law['act_raw_text']))

# Now add full_doc to each metadata chunk
for chunk in tqdm(metadata_chunks, desc="Adding full_doc"):
    celex = chunk['celex']
    if celex in celex_to_text:
        chunk['full_doc'] = celex_to_text[celex]
    else:
        chunk['full_doc'] = None  # or you can set to empty string: ''

# Save the updated metadata_chunks to a pickle file
with open('metadata_chunks_with_full_doc.pkl', 'wb') as f:
    pickle.dump(metadata_chunks, f)

print(f"Saved {len(metadata_chunks)} chunks with full_doc added")
````

````python
with open('metadata_chunks_with_full_doc.pkl','rb') as m:
    metadat_chunks = pickle.load(m)
````

````python
len(metadat_chunks)
````

````python
# Count chunks with and without full_doc
chunks_with_full_doc = sum(1 for chunk in metadat_chunks if 'full_doc' in chunk)
chunks_without_full_doc = len(metadat_chunks) - chunks_with_full_doc

print(f"Total chunks: {len(metadat_chunks)}")
print(f"Chunks with full_doc: {chunks_with_full_doc}")
print(f"Chunks without full_doc: {chunks_without_full_doc}")
````

````python
# Check if all chunks have the key
all_have_full_doc = all('full_doc' in chunk for chunk in metadat_chunks)
print(f"All chunks have full_doc: {all_have_full_doc}")

# If not all have it, find which ones are missing
if not all_have_full_doc:
    missing_indices = [i for i, chunk in enumerate(metadat_chunks) if 'full_doc' not in chunk]
    print(f"Found {len(missing_indices)} chunks missing full_doc")
    print(f"First few missing indices: {missing_indices[:5]}")
````

````python
from tqdm import tqdm

# Create lookup dictionary
celex_to_text_original = dict(zip(law['CELEX'], law['act_raw_text']))

# Verification counters
correct = 0
incorrect = 0
missing_in_original = 0
missing_in_chunk = 0
mismatches = []

# Check all chunks
for i, chunk in enumerate(tqdm(metadat_chunks, desc="Verifying chunks")):
    celex = chunk['celex']
    chunk_text = chunk.get('full_doc')
    
    # Check if CELEX exists in original dataframe
    if celex not in celex_to_text_original:
        missing_in_original += 1
        continue
    
    expected_text = celex_to_text_original[celex]
    
    # Check if chunk has full_doc
    if chunk_text is None:
        missing_in_chunk += 1
        mismatches.append({
            'index': i,
            'celex': celex,
            'issue': 'missing_full_doc'
        })
    # Check if text matches
    elif chunk_text != expected_text:
        incorrect += 1
        mismatches.append({
            'index': i,
            'celex': celex,
            'issue': 'text_mismatch',
            'chunk_preview': chunk_text[:200],
            'expected_preview': expected_text[:200]
        })
    else:
        correct += 1

# Print results
print("\n" + "="*50)
print("VERIFICATION RESULTS")
print("="*50)
print(f"Total chunks: {len(metadat_chunks)}")
print(f"✓ Correct matches: {correct}")
print(f"✗ Text mismatches: {incorrect}")
print(f"✗ Missing full_doc in chunk: {missing_in_chunk}")
print(f"✗ CELEX not in original dataframe: {missing_in_original}")
````
